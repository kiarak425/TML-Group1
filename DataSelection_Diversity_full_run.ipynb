```python
from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
import numpy as np
import json
```

```python
#------------Test run on subset of dataset-------------------------#

#Load our AceReason dataset
from datasets import load_dataset

dataset_url ="redsgnaoh/acereason11_100k"
full_dataset = load_dataset(dataset_url,split="train")

user_questions = full_dataset["instruction"][600:1000]


#Create embeddings for the user_questions
model = SentenceTransformer("all-MiniLM-L6-v2")

embeddings = model.encode(user_questions, batch_size=100)

#create clusters of the user_questions
kmeans = KMeans(n_clusters=10, random_state = 30)
cluster_labels = kmeans.fit_predict(embeddings)
sample_list = []
rng = np.random.default_rng(2)


#create a random list of samples
for cluster_id in range(kmeans.n_clusters):
  sample_cluster_indices = np.where(cluster_labels == cluster_id)[0]
  samples = rng.choice(sample_cluster_indices, size=5,replace=False)

  #test run
  print("Cluster ID: ",cluster_id)
  for idx in samples[:2]:
    print(user_questions[idx])
    print("- - - - - - - - - - - - - -")
  print("_______________________")

  sample_list.extend(samples)
```

```python
#------------Full 15k Run-------------------------#

#Load our AceReason dataset
from datasets import load_dataset

dataset_url ="redsgnaoh/acereason11_100k"
full_dataset = load_dataset(dataset_url,split="train")

user_questions = full_dataset["instruction"]


#Create embeddings for the user_questions
model = SentenceTransformer("all-MiniLM-L6-v2")

all_embeddings = model.encode(user_questions, batch_size=100,show_progress_bar=True)

np.save("/content/drive/MyDrive/all_embeddings.npy", all_embeddings)
```

```python
#create clusters of the user_questions
kmeans = KMeans(n_clusters=50, random_state = 30)
cluster_labels = kmeans.fit_predict(all_embeddings)

np.save("/content/drive/MyDrive/cluster_results", cluster_labels)
```

```python
#create clusters of the user_questions
kmeans = KMeans(n_clusters=100, random_state = 45)
cluster_labels = kmeans.fit_predict(all_embeddings)

np.save("/content/drive/MyDrive/100_cluster_results", cluster_labels)
```

```python
#for cluster size 50

cluster_size = np.bincount(cluster_labels)
for c,i in enumerate(cluster_size):
  print(c,i)
```

```python
#for cluster size 100

cluster_size_100 = np.bincount(cluster_labels)
for c,i in enumerate(cluster_size_100):
  print(c,i)
```

```python
sample_list = []
rng = np.random.default_rng(2)


#create a random list of samples
for cluster_id in range(kmeans.n_clusters):
  sample_cluster_indices = np.where(cluster_labels == cluster_id)[0]
  samples = rng.choice(sample_cluster_indices, size= 15000//kmeans.n_clusters,replace=False)

  sample_list.extend(samples)


diversity_dataset = full_dataset.select([int(idx) for idx in sample_list])

with open ("/content/drive/MyDrive/diversity_dataset_final.json","w") as f:
 json.dump(diversity_dataset.to_list(),f)
```
